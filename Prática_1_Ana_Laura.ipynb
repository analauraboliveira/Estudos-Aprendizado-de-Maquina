# BLOCO 3 — PRIMEIRO MODELO DE MACHINE LEARNING

```python
dados = df[[
    "Survived",
    "Pclass",
    "Sex",
    "Age",
    "SibSp",
    "Parch",
    "Fare"
]].copy()
```

```python
dados["Sex"] = (dados["Sex"] == "female").astype(int)
```

```python
dados.head()
```

```python
dados["Age"] = dados["Age"].fillna(dados["Age"].median())
```

```python
dados.isna().sum()
```

```python
X = dados.drop(columns="Survived")
y = dados["Survived"]
```

```python
X.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split

X_treino, X_teste, y_treino, y_teste = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)
```

```python
X_treino.shape
```

```python
X_teste.shape
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
```

```python
knn.fit(X_treino, y_treino)
```

```python
previsoes = knn.predict(X_teste)
```

```python
previsoes[:10]
```

```python
y_teste.values[:10]
```

> **Tarefa 10**

**Quantos dos dez primeiros exemplos foram classificados corretamente?**

Dos 10 primeiros exemplos do conjunto de teste, 6 foram classificados corretamente.

> **Tarefa 11**

```python
knn.score(X_teste, y_teste)
```

```python
(previsoes == y_teste.values).mean()
```

```python
knn.score(X_treino, y_treino)
```

**Resposta:** A acurácia obtida no conjunto de teste foi aproximadamente 67,91%. No conjunto de treinamento, a acurácia foi aproximadamente 80,90%.

```python
for k in [1, 3, 5, 7, 10]:
    modelo = KNeighborsClassifier(n_neighbors=k)
    modelo.fit(X_treino, y_treino)
    acuracia = modelo.score(X_teste, y_teste)
    print(k, acuracia)
```

> **Tarefa 12**

**Qual k apresentou a maior acurácia?**

k = 1, com aproximadamente 70,15%.

**Qual apresentou a menor?**

k = 3 e k = 7, ambos com aproximadamente 67,54%.
